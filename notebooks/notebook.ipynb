```python

## 1. Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

## 2. Create Fictional Data
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Sales': [150, 200, 250, 300, 400]
}
df = pd.DataFrame(data)

## 3. Display Data
df

## 4. Plot Sales Data
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```
